## 1. Mount Google Drive
Models (~GB) and built TensorRT engines are persisted here so they survive
session restarts instead of re-downloading / rebuilding every time.


```python
from google.colab import drive
drive.mount('/content/drive')

import os
PERSIST = '/content/drive/MyDrive/VisoMaster'
MODELS_DIR = os.path.join(PERSIST, 'model_assets')
TRT_CACHE  = os.path.join(PERSIST, 'tensorrt-engines')
os.makedirs(MODELS_DIR, exist_ok=True)
os.makedirs(TRT_CACHE, exist_ok=True)
# Point the engine at the persistent Drive locations.
os.environ['VISOMASTER_MODELS_DIR'] = MODELS_DIR
os.environ['VISOMASTER_TRT_CACHE']  = TRT_CACHE
```
